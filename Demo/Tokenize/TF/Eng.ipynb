```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
documents = [
    "The cat in the hat",
    "The dog in the fog",
    "The cat on the mat",
    "A dog is a man's best friend"
]
```

```python
from nltk.tokenize import word_tokenize
```

```python
import nltk
from nltk.corpus import stopwords

# Tải stopwords từ NLTK nếu chưa tải
# nltk.download('stopwords')

# Sử dụng danh sách từ dừng (stopwords) của tiếng Anh từ NLTK
stopword_list = list(stopwords.words('english'))
stopwords = [word for word in stopword_list if word.strip() != '']

def tokenize(sentence):
    return word_tokenize(sentence)
```

```python
# Khởi tạo TfidfVectorizer
vectorizer = TfidfVectorizer(stop_words = stopwords,
    tokenizer = tokenize)

# Chuyển đổi các câu thành ma trận TF-IDF
X = vectorizer.fit_transform(documents)

# Chuyển ma trận thành DataFrame để dễ dàng xem
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())

# In kết quả
print(df)
```
