```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
import string
from pyvi import ViTokenizer
```

```python
data = [
    "Đội ngũ bảo mật giám sát lưu lượng mạng.",
    "Quản trị viên hệ thống vá lỗi phần mềm định kỳ.",
    "Phần mềm mã hóa dữ liệu nhạy cảm để bảo vệ quyền riêng tư.",
    "Tường lửa giúp bảo vệ mạng khỏi các tấn công xâm nhập.",
    "Sao lưu định kỳ giúp khôi phục dữ liệu sau các sự cố bảo mật.",
    "Bảo mật hệ thống là một phần không thể thiếu trong việc quản lý rủi ro.",
    "Xâm nhập trái phép có thể gây ra hậu quả nghiêm trọng nếu không được phát hiện kịp thời.",
    "Các cuộc tấn công mạng có thể làm gián đoạn hoạt động của doanh nghiệp.",
    "Mã độc là mối nguy hiểm lớn đối với các hệ thống thông tin."
]
```

```python
stopwords = ["ai",'ai_ai', 'ai_nấy', 'ai_đó', 'bao_lâu', 'bao_nhiêu', 'biết', 'biết_bao', 'biết_chắc', "và","với"]
```

```python
def tokenizerVN(text):
    return ViTokenizer.tokenize(text)

# Định nghĩa tokenize
def tokenize(sentence):
    # Tokenize và loại bỏ dấu câu
    tokens = tokenizerVN(sentence).split()
    tokens = [word for word in tokens if word not in string.punctuation]  # Loại bỏ dấu câu
    return tokens
```

```python
# Khởi tạo TfidfVectorizer
# Tạo CountVectorizer
vectorizerCV = TfidfVectorizer(
    stop_words=stopwords,
    tokenizer=tokenize,
)
# Set pandas options to display full data
pd.set_option('display.max_columns', None)  # Show all columns
pd.set_option('display.max_rows', None)  # Show all rows
pd.set_option('display.width', None)  # Auto-detect width
pd.set_option('display.max_colwidth', None)  # Show full content of each column
# Chuyển đổi các câu thành ma trận TF-IDF
X = vectorizerCV.fit_transform(data)

# Chuyển ma trận thành DataFrame để dễ dàng xem
df = pd.DataFrame(X.toarray(), columns=vectorizerCV.get_feature_names_out())

# In kết quả
print(df)
```
